##Exercise 3.2: Counting with Loops   

```python
predictions = [0, 1, 1, 0, 1, 0, 1, 1]
true_labels = [0, 1, 0, 0, 1, 0, 1, 1]

# Count correct predictions
# Store in correct_count
correct_count = 0
# Loop through all indices of the lists
for i in range(len(predictions)):
    if predictions[i] == true_labels[i]:
        # Increment the counter by 1 if they match
        correct_count += 1


print(f"Correct: {correct_count}/{len(predictions)}")
print(f"Accuracy: {correct_count/len(predictions)*100:.1f}%")
```

##Exercise 4.1: Mean Function   

```python
def calculate_mean(values):
    """
    Calculate mean (average) of a list.

    Args:
        values: List of numbers

    Returns:
        Mean as float
    """

    return sum(values) / len(values)

test = [10, 20, 30, 40, 50]
result = calculate_mean(test)
print(f"Mean of {test}: {result}")
```

##Exercise 4.2: Distance Function

```python
def euclidean_distance(point1, point2):
    """
    Calculate Euclidean distance between two 2D points.

    Formula: sqrt((x1-x2)² + (y1-y2)²)
    """
    x1, y1 = point1
    x2, y2 = point2
    return ((x1 - x2) ** 2 + (y1 - y2) ** 2) ** 0.5

p1, p2 = [0, 0], [3, 4]
dist = euclidean_distance(p1, p2)
print(f"Distance: {dist}")
```

##Exercise 4.3: Normalization Function   

```python
def normalize_list(values):
    """
    Normalize to [0,1] using min-max scaling.

    Formula: (x - min) / (max - min)
    """
    min_val = min(values)
    max_val = max(values)
    normalized = []
    for x in values:
        normalized.append((x - min_val) / (max_val - min_val))
    return normalized


# Test
data = [10, 20, 30, 40, 50]
result = normalize_list(data)
print(f"Normalized: {[f'{x:.2f}' for x in result]}")
```

##Exercise 4.4: Sigmoid Function   

```python
def simple_sigmoid(x):
    """
    Sigmoid: σ(x) = 1 / (1 + e^(-x))
    """
    e = 2.718281828459045
    return 1 / (1 + e ** (-x))



# Test
for x in [-5, 0, 5]:
    print(f"sigmoid({x}) = {simple_sigmoid(x):.4f}")
```

##Final Challenge: Complete ML Pipeline without a model   

```python
# Given data
data_points = [1.5, 4.2, 2.8, 5.6, 3.1]
true_labels = [0, 1, 0, 1, 1]

# Task 1: Make predictions (if > 3.0, predict 1, else 0)
predictions = []
for x in data_points:
    if x > 3.0:
        predictions.append(1)
    else:
        predictions.append(0)

# Task 2: Calculate accuracy
correct = 0
for i in range(len(predictions)):
    if predictions[i] == true_labels[i]:
        correct += 1
accuracy = (correct / len(predictions)) * 100

# Task 3: Normalize data using min-max scaling
normalized_data = normalize_list(data_points)

# Results
print("╔═══════════════════════════════╗")
print("║   ML Pipeline Results        ║")
print("╚═══════════════════════════════╝")
print(f"\nData: {data_points}")
print(f"Normalized: {[f'{x:.3f}' for x in normalized_data]}")
print(f"\nTrue: {true_labels}")
print(f"Pred: {predictions}")
print(f"\nAccuracy: {accuracy:.1f}%")
```
